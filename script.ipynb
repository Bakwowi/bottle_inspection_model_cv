```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, img_as_float
import cv2
```

```python
img = img_as_float(io.imread('test.png', as_gray=True))
img
```

```python
kernel = np.ones((5, 5), np.float32) / 25
kernel
```

```python
gaussion_kernel = np.array([[1/16, 1/8, 1/16],
                            [1/8, 1/4, 1/8],
                            [1/16, 1/8, 1/16]])
gaussion_kernel
```

```python
conv_img_cv2 = cv2.filter2D(img, -1, kernel, borderType=cv2.BORDER_CONSTANT)
conv_img_cv2_gaussian = cv2.filter2D(img, -1, gaussion_kernel, borderType=cv2.BORDER_CONSTANT)
```

```python
cv2.imshow('Original Image', img)
cv2.imshow('Convolution with Kernel', conv_img_cv2)
cv2.imshow('Convolution with Gaussian Kernel', conv_img_cv2_gaussian)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
